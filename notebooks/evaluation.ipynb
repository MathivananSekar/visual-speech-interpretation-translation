```python
import os
def inspect_npz_files(speaker_ids, base_path):
    all_npz_paths = []
    for spk_id in speaker_ids:
        dir_spk = os.path.join(base_path, "processed", spk_id)
        for f in os.listdir(dir_spk):
            if f.endswith(".npz"):
                all_npz_paths.append(os.path.join(dir_spk, f))

    for path in all_npz_paths:
        data = np.load(path, allow_pickle=True)
        frames = data["frames"]
        print(f"File: {path}, Shape: {frames.shape}")
        if frames.ndim != 4 or frames.shape[-1] != 3 or frames.shape[1:3] != (64, 128):
            print(f"ERROR: Invalid shape in {path}")

# Run this before training
speaker_ids = ["s1", "s2", "s3", "s4", "s5", "s6", "s7", "s8", "s9", "s10"]  # Replace with actual IDs
base_path = "../data"       # Replace with actual path
inspect_npz_files(speaker_ids, base_path)
```
